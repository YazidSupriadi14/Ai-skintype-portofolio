**Import Library**

````python
import os
import random
import torch
import numpy as np
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
from PIL import Image
from tqdm import tqdm
from sklearn.metrics import confusion_matrix
from torch.utils.data import DataLoader
from torchvision import datasets, transforms, models
import seaborn as sns
import gradio as gr

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    BartTokenizer,
    BartForConditionalGeneration
)

from huggingface_hub import create_repo, upload_folder
````

````python
# Download dataset from Kaggle
dataset_path = kagglehub.dataset_download("manithj/oily-and-dry-skin-dataset")
print("Path to dataset files:", dataset_path)
````

**Data Preparation**

````python
base_path = "/kaggle/input/oily-and-dry-skin-dataset/Oily-Dry-Skin-Dataset"

# Transformations to preprocess images before feeding them into the model
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

# Define paths for training, validation, and testing folders
train_dir = os.path.join(base_path, "train")
val_dir = os.path.join(base_path, "valid")
test_dir = os.path.join(base_path, "test")

# Load datasets from folders, each subfolder represents a class label
train_dataset = datasets.ImageFolder(train_dir, transform=transform)
val_dataset = datasets.ImageFolder(val_dir, transform=transform)
test_dataset = datasets.ImageFolder(test_dir, transform=transform)

# Create data loaders for batch-wise iteration during training and evaluation
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32)
test_loader = DataLoader(test_dataset, batch_size=32)

class_names = train_dataset.classes  # ['dry', 'normal', 'oily']
print("Classes:", class_names)
````

**Model Preparation with restnet18**

````python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load a pretrained ResNet18 model from torchvision
model = models.resnet18(pretrained=True)
num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, 3)
model = model.to(device)
````

````python
class_names = train_dataset.classes
print(class_names)
````

**Training Phase**

````python
# Define the loss function (CrossEntropyLoss for multi-class classification)
criterion = nn.CrossEntropyLoss()

# Set up the optimizer (Adam optimizer with learning rate of 0.001)
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Define the training function
def train(model, loader, optimizer, criterion, epoch):
    model.train()
    total, correct, total_loss = 0, 0, 0

    # Use tqdm to display a progress bar during the epoch
    loop = tqdm(loader, desc=f"Epoch {epoch+1}")

    # Loop over batches in the data loader
    for images, labels in loop:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        _, preds = torch.max(outputs, 1)
        correct += (preds == labels).sum().item()
        total += labels.size(0)

        loop.set_postfix(loss=loss.item(), acc=100*correct/total)

    # Return the average loss and accuracy for the epoch
    return total_loss / len(loader), 100 * correct / total
````

````python
# Initialize an empty list to store metrics for each epoch
metrics = []

# Set the number of epochs to train the model
num_epochs = 12

# Loop over the range of epochs
for epoch in range(num_epochs):
    loss, acc = train(model, train_loader, optimizer, criterion, epoch)

    # Append the results for this epoch (epoch number, accuracy, loss) to the metrics list
    metrics.append({
        "epoch": epoch + 1,
        "accuracy": acc,
        "loss": loss
    })
    print(f"[Epoch {epoch+1}] Loss: {loss:.4f} | Accuracy: {acc:.2f}%")
````

**Evaluation with confusion matrix**

````python
# Set device to GPU if available, else CPU
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Define a function to plot confusion matrix
def plot_confusion_matrix(preds, labels, class_names):
    cm = confusion_matrix(labels, preds)
    cm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]  # Normalisasi
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='.2f', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
    plt.xlabel('Predicted')
    plt.ylabel('True')
    plt.title('Confusion Matrix')
    plt.show()

# Define a function for evaluating the model on the validation set
def evaluate(model, loader, criterion, class_names):
    model.eval()
    all_preds = []
    all_labels = []

    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)

            # Making Prediction
            outputs = model(images)
            _, preds = torch.max(outputs, 1)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    # Plot Confusion Matrix
    plot_confusion_matrix(all_preds, all_labels, class_names)

    # Calculate accuracy
    correct = np.sum(np.array(all_preds) == np.array(all_labels))
    total = len(all_labels)
    accuracy = 100 * correct / total
    return accuracy


model.to(device)
class_names = ['dry', 'oily', 'normal']
val_accuracy = evaluate(model, val_loader, criterion, class_names)

#Print validation Accuracy
print(f"Validation Accuracy: {val_accuracy:.2f}%")
````

**Scrap Data from internet with BeautifulSoap**

````python
# Function to scrape article from URL
def scrape_article(url):
    # Define headers to avoid blocking by the website (pretend to be a browser)
    headers = {"User-Agent": "Mozilla/5.0"}

    try:
        # Send a GET request to the URL with the defined headers
        res = requests.get(url, headers=headers, timeout=10)

        # Parse the HTML content of the page using BeautifulSoup
        soup = BeautifulSoup(res.text, "html.parser")

        # Extract the title of the article from the <title> tag
        title = soup.title.get_text(strip=True) if soup.title else "No Title"

        # Extract all paragraphs <p> from the article and join them together
        paragraphs = soup.find_all("p")
        content = "\n".join(p.get_text(strip=True) for p in paragraphs if p.get_text(strip=True))

        # Check if the article content is long enough (more than 100 characters)
        if len(content) < 100:
            return "Konten terlalu pendek atau tidak valid."

        # Return a dictionary with the article title, URL, and content
        return {
            "title": title,
            "url": url,
            "content": content
        }
    except Exception as e:
        return f"Error scraping {url}: {str(e)}"

# Function to save article to a file
def save_article(article, folder="/mnt/data/skin_articles"):
    os.makedirs(folder, exist_ok=True)
    fname = f"{folder}/{article['title'][:50].replace('/', '-')}.txt"
    with open(fname, "w", encoding="utf-8") as f:
        f.write(f"{article['title']}\n{article['url']}\n\n{article['content']}")

# URLs of articles categorized by skin type
urls = {
    "oily": [
        "https://blog.avoskinbeauty.com/produk-avoskin-untuk-kulit-berminyak/",
        "https://blog.avoskinbeauty.com/rekomendasi-moisturizer-untuk-kulit-berminyak-dan-berjerawat/"
    ],
    "dry": [
        "https://blog.avoskinbeauty.com/mengatasi-kulit-kering-dengan-produk-avoskin/",
        "https://blog.avoskinbeauty.com/skincare-untuk-kulit-kering-avoskin-terbaru-2023/"
    ],
    "normal": [
        "https://readmore.id/cara-merawat-wajah-normal/"
    ]
}

# Scraping and saving articles for each skin type
for skin_type, url_list in urls.items():
    for url in url_list:
        article = scrape_article(url)
        if isinstance(article, dict):
            save_article(article, folder=f"/content/skin_articles/{skin_type}")
            print(f"Artikel '{article['title']}' untuk jenis kulit {skin_type} berhasil disimpan.")
        else:
            print(f"Error saat scraping {url}: {article}")
````

**Inference using gradio with scrapping data from internet for recomendation using cahya/bert2gpt-indonesian-summarization for summarization**

````python

# Load class skin types
class_names = ['dry', 'normal', 'oily']

# Load ResNet model for skin type classification
model_resnet = models.resnet18()
model_resnet.fc = torch.nn.Linear(model_resnet.fc.in_features, len(class_names))
model_resnet.load_state_dict(torch.load("/content/yazidsupriadi/resnet18-skin-type/pytorch_model.bin", map_location=torch.device("cpu")))
model_resnet.eval()

# Image transform
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

# Summarization model
# Load tokenizer dan model
from transformers import BertTokenizer, EncoderDecoderModel

# Load tokenizer dan model
tokenizer = BertTokenizer.from_pretrained("cahya/bert2gpt-indonesian-summarization")
tokenizer.bos_token = tokenizer.cls_token
tokenizer.eos_token = tokenizer.sep_token
model = EncoderDecoderModel.from_pretrained("cahya/bert2gpt-indonesian-summarization")

def summarize_text(text):
    inputs = tokenizer.encode(text, return_tensors='pt', max_length=512, truncation=True)
    summary_ids = model.generate(
        inputs,
        max_length=150,
        min_length=50,
        num_beams=4,
        length_penalty=2.0,
        early_stopping=True
    )
    return tokenizer.decode(summary_ids[0], skip_special_tokens=True)

# Prediction function
def predict(image):
    try:
        image = image.convert("RGB")
        img_tensor = transform(image).unsqueeze(0)
        with torch.no_grad():
            outputs = model_resnet(img_tensor)
            _, predicted = torch.max(outputs, 1)
            label = class_names[predicted.item()]
            confidence = torch.nn.functional.softmax(outputs, dim=1)[0][predicted.item()].item()

        # Load and summarize article
        folder_path = os.path.join("/content/skin_articles", label)
        article_files = [f for f in os.listdir(folder_path) if f.endswith(".txt")]
        if not article_files:
            return f"Tidak ada artikel untuk tipe kulit {label}."

        selected_file = random.choice(article_files)
        with open(os.path.join(folder_path, selected_file), "r", encoding="utf-8") as f:
            content = f.read()

        summarized = summarize_text(content)

        return f"""🧠 **Tipe Kulit Terdeteksi**: `{label.capitalize()}` ({confidence*100:.2f}% confidence)

📚 **Rekomendasi Perawatan (Ringkasan)**:
{summarized}
"""
    except Exception as e:
        return f"⚠️ Error: {e}"

# Launch Gradio UI
interface = gr.Interface(
    fn=predict,
    inputs=gr.Image(type="pil"),
    outputs=gr.Markdown(),
    title="Deteksi Tipe Kulit & Rekomendasi Perawatan",
    description="Unggah foto wajah untuk klasifikasi tipe kulit (dry/normal/oily) dan dapatkan rekomendasi artikel yang diringkas otomatis."
)

interface.launch(debug=True)
````
